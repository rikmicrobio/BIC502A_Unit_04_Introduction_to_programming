# 3. Input, Output and f-strings

`input()` accepts user input and returns a string. Convert it when numeric input is needed.

An **f-string** is a formatted string literal. Put `f` before the string and variables or expressions inside `{}`.


````python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
print(f"Name: {name}; Age: {age}")
````

### Why the `f`?
Without `f`, `{x}` is treated as ordinary text. With `f`, Python evaluates the expression inside the braces.


````python
x = 10
print("Without f:", "The value of x is {x}")
print("With f:", f"The value of x is {x}")
````

# 5. Conditional Statements

An `if` statement executes a block only when its condition is true. Python uses indentation to define the block.


````python
temperature = 37
if temperature == 37:
    print("Temperature is 37 degrees")
````

## 5.1 `if-else`

`else` provides the alternative branch when the `if` condition is false.


````python
temperature = 39
if temperature <= 37:
    print("Normal temperature")
else:
    print("High temperature")
````

## 5.2 `if-elif-else`

Use `elif` to test additional conditions. Python executes the first matching branch.


````python
score = 78
if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Grade D")
````

## 5.3 Logical operators
- `and`: both conditions must be true.
- `or`: at least one condition must be true.
- `not`: reverses a Boolean value.


````python
age = 22
has_sample = True
if age >= 18 and has_sample:
    print("Eligible")

organism = "E. coli"
if organism == "E. coli" or organism == "Klebsiella":
    print("Selected organism")

contaminated = False
if not contaminated:
    print("Sample can proceed")
````

## 5.4 Nested conditions

A nested conditional is a conditional inside another conditional. Use it when one decision depends on an earlier decision.


````python
sample_available = True
concentration = 15

if sample_available:
    if concentration >= 10:
        print("Sample accepted")
    else:
        print("Concentration too low")
else:
    print("Sample unavailable")
````

# 6. Functions

A **function** is a reusable block of code designed to perform a specific task. Functions reduce repetition and make programs easier to organize, test and maintain.

Basic structure:
```python
def function_name():
    statements
```

Defining a function does not execute it; calling the function does.


````python
def greet():
    print("Hello, Python!")

greet()
greet()
````

## 6.1 Parameters and arguments
A **parameter** is a placeholder in a function definition. An **argument** is the actual value supplied when the function is called.


````python
def greet(name):
    print(f"Hello, {name}!")

greet("Rik")
greet("Student")
````

## 6.2 Multiple parameters


````python
def add(a, b):
    print(a + b)

add(10, 20)
add(5, 7)
````

## 6.3 `return`

`print()` displays a value. `return` sends a value back to the code that called the function. Returned values can be stored, compared or used in further calculations.


````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

## 6.4 `print()` versus `return`


````python
def square_print(x):
    print(x * x)

def square_return(x):
    return x * x

square_print(5)
result = square_return(5)
print(result)
````

## 6.5 Biological function example


````python
def calculate_gc_content(g, c, length):
    gc = ((g + c) / length) * 100
    return gc

gc = calculate_gc_content(30, 20, 100)
print(f"GC content = {gc}%")
````

## 6.6 Functions containing conditions


````python
def classify_temperature(temp):
    if temp <= 37:
        return "Normal"
    else:
        return "High"

print(classify_temperature(39))
````

## 6.7 Default and keyword arguments

A default parameter is used when no argument is supplied. Keyword arguments are passed by parameter name.


````python
def greet(name="Student"):
    print(f"Hello, {name}!")

greet()
greet("Rik")

def student_info(name, age):
    print(f"Name: {name}; Age: {age}")

student_info(age=25, name="Rik")
````

# 7. Functions Calling Functions

A function can use another function. This lets us break larger problems into smaller tasks.


````python
def square(x):
    return x * x

def sum_of_squares(a, b):
    return square(a) + square(b)

print(sum_of_squares(3, 4))
````

# 8. Scope

Scope describes where a variable name can be accessed. Introduce scope **after functions**, because local and global variables make more sense once students understand function execution.

The two beginner concepts are **local scope** and **global scope**.


## 8.1 Local scope
A variable created inside a function normally belongs to that function's local scope.


````python
def calculate():
    result = 10 * 2
    print(result)

calculate()
````

`result` cannot normally be accessed outside `calculate()` because it is local to that function.


## 8.2 Global scope
A variable created outside a function is normally in global scope.


````python
organism = "E. coli"

def show_organism():
    print(organism)

show_organism()
````

## 8.3 Local and global variables with the same name

A local variable can have the same name as a global variable. They are separate variables in different scopes.


````python
x = 1

def foo():
    x = 11
    y = 12
    print(f"foo sees x equal to {x}")
    print(f"foo sees y equal to {y}")

foo()
print(f"global sees x equal to {x}")
````

Expected output:
```text
foo sees x equal to 11
foo sees y equal to 12
global sees x equal to 1
```
The `x` inside `foo()` is local and does not replace the global `x`.


## 8.4 The `global` keyword

Use `global` when a function needs to assign a new value to a module-level variable. It should be used sparingly.


````python
x = 10

def change_x():
    global x
    x = 20

change_x()
print(x)
````

# 9. `match-case`

Python 3.10 introduced structural pattern matching. `match-case` can be used for many switch-like tasks where one value is compared with several cases.


````python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Other day")
````

`_` is the catch-all case.

Use `if-elif` when you need flexible logical conditions or ranges. Use `match-case` when matching values or patterns is clearer.


# 10. Integrated Example

This example combines variables, conditions, functions, `return` and f-strings.


````python
def classify_sample(ph, temperature, concentration):
    if ph < 7 or ph > 7.5:
        return "Adjust pH"
    if temperature != 37:
        return "Check temperature"
    if concentration < 10:
        return "Concentration too low"
    return "Sample accepted"

ph = 7.2
temperature = 37
concentration = 15
result = classify_sample(ph, temperature, concentration)
print(f"Result: {result}")
````

# 11. Common Beginner Errors

1. `=` is assignment; `==` is comparison.
2. Indentation defines blocks in Python.
3. Do not forget `:` after `if`, `elif`, `else`, `def`, `match` and `case`.
4. `input()` returns a string unless converted.
5. `print()` displays; `return` sends a value back.
6. Variables created inside functions are normally local.
7. `match-case` requires Python 3.10 or later.
