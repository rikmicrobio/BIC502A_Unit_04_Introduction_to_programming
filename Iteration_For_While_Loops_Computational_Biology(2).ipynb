# Python Programming: Iteration, `for` Loops and `while` Loops
### A for loop is a programming control flow statement that repeats a block of code a specific number of times or across a sequence of items.

### A while loop is a programming control structure that repeatedly executes a block of code as long as a specified condition evaluates to true.

## 1. What is iteration?

**Iteration** means repeating a set of instructions. A loop lets us repeat code without writing the same instruction many times.

For example, suppose we want to print each nucleotide in a DNA sequence. We could print each character separately, but a loop can do this for us.

```python
print("A")
print("T")
print("G")
print("C")
```

The loop-based version is shorter and works for sequences of different lengths.


````python
dna = "ATGC"

for nucleotide in dna:
    print(nucleotide)
````

### What happens during each iteration?

| Iteration | Value of `nucleotide` | Action |
|---:|---|---|
| 1 | `"A"` | Print A |
| 2 | `"T"` | Print T |
| 3 | `"G"` | Print G |
| 4 | `"C"` | Print C |

The loop takes one character at a time from the string. After all characters have been processed, the loop ends.


## 2. The `for` loop

Use a `for` loop when you want to process items in a sequence or repeat an operation a set number of times.

### General structure

```python
for item in sequence:
    # Instructions to repeat
```

- `for` starts the loop.
- `item` receives the next value on each iteration.
- `in` connects the variable to the sequence or iterable.
- The indented statements form the loop body.


````python
for number in range(1, 6):
    print(number)
````

### Understanding `range()`

- `range(5)` produces `0, 1, 2, 3, 4`.
- `range(1, 6)` produces `1, 2, 3, 4, 5`.
- `range(2, 11, 2)` produces `2, 4, 6, 8, 10`.

**Important:** The stop value is excluded. In `range(1, 6)`, the number 6 is not included.


````python
# Example: count the nucleotides in a DNA sequence using a for loop

dna = "ATGCGCTA"
count = 0

for nucleotide in dna:
    count = count + 1

print("Sequence length:", count)
````

## 3. The `while` loop

A `while` loop repeats its body **as long as its condition is true**.

### General structure

```python
while condition:
    # Instructions to repeat
    # Update values so the loop can eventually stop
```

A `while` loop commonly needs:
1. An initial value.
2. A condition to check.
3. An update that changes the value involved in the condition.

### Example: print numbers from 1 to 5


````python
number = 1

while number <= 12:
    print(number)
    number = number + 1
````

### Trace the loop: means manually following the execution of a loop step-by-step, just like a computer would, to track how the variables change over time.

| Current value of `number` | Is `number <= 5` true? | What happens? |
|---:|---|---|
| 1 | Yes | Print 1, then increase to 2 |
| 2 | Yes | Print 2, then increase to 3 |
| 3 | Yes | Print 3, then increase to 4 |
| 4 | Yes | Print 4, then increase to 5 |
| 5 | Yes | Print 5, then increase to 6 |
| 6 | No | Stop the loop |

### Watch out: infinite loops

If the value used in a `while` condition never changes, the condition may stay true forever. Always check that your loop can reach a stopping condition.


## 4. `for` loop vs `while` loop

| `for` loop | `while` loop |
|---|---|
| Iterates through an iterable, such as a string or list. | Repeats while a condition remains true. |
| Useful for processing every item in a sequence. | Useful when the number of repetitions depends on a condition. |
| Iteration over the iterable is handled automatically. | You often need to update a counter or state variable yourself. |

Both loops can solve many of the same problems. Choose the one that makes the task clearest.


---
1. Iterating an iterable vs. Checking a condition

• for loop: Automatically looks at every single item inside a collection (like a list of names).
---
• while loop: Keeps running as long as a mathematical or logical condition remains true.
