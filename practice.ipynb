# 📘 Session 1: Iteration and the `while` loop

## 1.1 What is iteration?

**Iteration** means *repeating* a set of instructions. Computers are excellent at this; humans get bored and make mistakes.

Think of real lab situations:
- 🧪 Pipette into each of 96 wells → repeat 96 times
- 🔬 Keep diluting a sample **until** the concentration drops below a threshold
- 📄 Read lines from a file **until** the file ends

Every loop has two ingredients:
1. **An action** to repeat (the *loop body*)
2. **A rule for when to stop** (the *condition*)

Without a stopping rule, the loop runs forever. That is called an **infinite loop**. 😬

## 1.2 Without a loop (the painful way)
Suppose we want to print the numbers 1 to 5.

````python
print(1)
print(2)
print(3)
print(4)
print(5)
````

This works for 5 numbers, but what about 5,000? We need a loop.

## 1.3 The `while` loop

```python
while condition:
    # indented code that repeats
    # as long as the condition is True
```

**Reading it aloud:** *"While this is true, keep doing the following."*

⚠️ **Indentation matters!** In Python, the indented lines are the loop body. Use 4 spaces (or Tab).

````python
count = 1                 # 1. Initialise a variable
while count <= 5:         # 2. Check the condition
    print(count)          # 3. Do the action
    count = count + 1     # 4. Update the variable (so we eventually stop!)

print("Loop finished!")   # Not indented -> runs once, after the loop
````

### 🔍 Tracing the loop (the most important skill!)
A **trace table** shows the value of each variable at each pass. Do this on paper whenever a loop confuses you.

| Pass | `count` at start | Condition `count <= 5` | Action | `count` after |
|---|---|---|---|---|
| 1 | 1 | True | print 1 | 2 |
| 2 | 2 | True | print 2 | 3 |
| 3 | 3 | True | print 3 | 4 |
| 4 | 4 | True | print 4 | 5 |
| 5 | 5 | True | print 5 | 6 |
| 6 | 6 | **False** | *(loop exits)* | 6 |

### The 4 parts of every `while` loop
1. **Initialise** the variable(s)
2. **Condition** to check
3. **Body** (the work)
4. **Update** so the condition can eventually become False

## 1.4 Shorthand: `+=`
`count = count + 1` is so common that Python has a shortcut:

| Long form | Short form |
|---|---|
| `x = x + 1` | `x += 1` |
| `x = x - 1` | `x -= 1` |
| `x = x * 2` | `x *= 2` |
| `x = x / 2` | `x /= 2` |

````python
# Countdown
n = 10
while n > 0:
    print(n, end=" ")   # end=" " keeps output on one line
    n -= 1
print("loop ends here")
````

## 1.5 A biological example: bacterial growth
A culture starts with **100 cells** and **doubles every generation**. How many generations until it exceeds **1 million** cells?

We don't know in advance how many repetitions are needed, so `while` is the natural choice.

````python
cells = 100
generation = 0

while cells <= 1_000_000:     # underscores just make big numbers readable
    cells *= 2
    generation += 1
    print(f"Generation {generation}: {cells:,} cells")

print(f"\nIt took {generation} generations to exceed 1 million cells.")
````

> 📝 **New syntax:** `f"..."` is an *f-string*. Anything inside `{ }` is replaced by its value. `:,` adds thousands separators.

## 1.6 ⚠️ The infinite loop
What if we **forget the update step**?

```python
count = 1
while count <= 5:
    print(count)
    # forgot count += 1  →  count is always 1  →  condition is always True  →  runs forever!
```

**I deliberately won't run that.** If it ever happens to you in Colab, press the ⏹️ **stop button** on the cell.

Below is a *safe* version that demonstrates the bug with a safety limit:

````python
count = 1
safety = 0
while count <= 5:
    print("count is still", count)
    safety += 1
    if safety == 4:
        print("... this would go on forever! Stopping manually.")
        break
````

## 1.7 `break` and `continue`: controlling the flow

| Keyword | What it does |
|---|---|
| `break` | Exit the loop **immediately** |
| `continue` | Skip the rest of **this pass** and jump to the next one |

````python
# break: stop as soon as we find what we want
n = 1
while True:                 # "forever" loop, but we break out of it
    if n % 7 == 0:          # % is the remainder operator: n % 7 == 0 means n is divisible by 7
        print("First multiple of 7 is", n)
        break
    n += 1
````

````python
# continue: skip certain values
n = 0
while n < 10:
    n += 1
    if n % 2 == 0:          # if n is even...
        continue            # ...skip the print and go straight to the next pass
    print(n, end=" ")
````

# 📗 Session 2: The `for` loop

## 2.1 `while` vs `for`

| Use `while` when... | Use `for` when... |
|---|---|
| You know **when to stop** but not how many repeats | You know **what to loop over** (a range, a string, a list...) |
| *"Keep going until the population is big enough"* | *"Do this for each of the 96 wells"* |

A `for` loop **handles the counting and updating for you**, so there are fewer places to make mistakes.

```python
for variable in sequence:
    # body runs once for each item in the sequence
```

## 2.2 Looping with `range()`
`range()` generates a sequence of numbers.

| Call | Produces |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -1)` | 10, 9, 8, ... 1 |

🔑 **The stop value is NOT included.** `range(5)` starts at **0** and ends at **4**. This is the source of many *off-by-one errors*.

````python
for i in range(5):
    print("i is", i)
````

````python
for i in range(2, 11, 2):     # start, stop, step
    print(i, end=" ")
````

````python
# Same job as our while-loop countdown, but shorter
for n in range(10, 0, -1):
    print(n, end=" ")
print("🚀")
````

### Convert `while` ↔ `for`
These two do exactly the same thing:

````python
# while version
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()

# for version
for i in range(1, 6):         # remember: stop value 6 is excluded
    print(i, end=" ")
````

## 2.3 Looping over strings
A string is a **sequence of characters**, so we can loop over it directly. This is perfect for DNA!

````python
dna = "ATGCGTA"
for base in dna:
    print(base)
````

````python
# Loop with an index too
for i in range(len(dna)):     # len() gives the length of the string
    print(f"Position {i}: {dna[i]}")
````

## 2.4 The accumulator pattern 🧮
One of the **most common patterns** in programming:

1. Create a variable **before** the loop (the accumulator)
2. **Update** it inside the loop
3. Use the result **after** the loop

| Goal | Start value | Update |
|---|---|---|
| Sum | `total = 0` | `total += x` |
| Count | `count = 0` | `count += 1` |
| Product | `result = 1` | `result *= x` |
| Max | `biggest = first value` | `if x > biggest: biggest = x` |

````python
# Sum of 1..100 (the famous Gauss problem)
total = 0
for n in range(1, 101):
    total += n
print("Sum of 1 to 100 =", total)
````

````python
# Count how many times 'A' appears in a DNA sequence
dna = "ATGCGATACGCTTAGGA"
a_count = 0
for base in dna:
    if base == "A":
        a_count += 1
print("Number of A's:", a_count)
````

````python
# Calculate GC content (percentage of G and C bases)
gc_count = 0
for base in dna:
    if base == "G" or base == "C":
        gc_count += 1

gc_percent = gc_count / len(dna) * 100
print(f"GC content: {gc_percent:.1f}%")
````

## 2.5 Conditions inside loops: `if / elif / else`
A loop plus an `if` lets us make decisions for each item.

```python
if condition1:
    ...
elif condition2:
    ...
else:
    ...
```

````python
# Classify each number
for n in range(1, 11):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print(n, "→ divisible by 3")
    elif n % 5 == 0:
        print(n, "→ divisible by 5")
    else:
        print(n)
````

## 2.6 Nested loops: loops inside loops 🪆
For **each** pass of the outer loop, the inner loop runs **completely**.

A perfect analogy: a **96-well plate** has 8 rows (A-H) and 12 columns (1-12).

````python
rows = "ABCDEFGH"
for row in rows:                 # outer loop: 8 passes
    for col in range(1, 13):     # inner loop: 12 passes each
        print(f"{row}{col}", end=" ")
    print()                      # new line after each row
````

````python
# Multiplication table
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:3}", end=" ")
    print()
````

````python
# Triangle pattern
for i in range(1, 6):
    print("*" * i)
````

# 📙 Session 3: Lists

## 3.1 Why do we need lists?
Imagine storing the molecular weights of 50 drugs:

```python
mw1 = 180.2
mw2 = 342.3
mw3 = 151.2
# ... 47 more variables?!
```

That's unmanageable. A **list** stores many values in **one variable**, in order.

```python
molecular_weights = [180.2, 342.3, 151.2]
```

````python
molecular_weights = [180.2, 342.3, 151.2, 780.9, 46.1]
drug_names = ["Aspirin", "Penicillin", "Paracetamol", "Rifampicin", "Ethanol"]

print(molecular_weights)
print(type(molecular_weights))
print("Number of items:", len(molecular_weights))
````

Lists can hold **any type** (even mixed), but it's good practice to keep one type per list.

````python
mixed = ["Aspirin", 180.2, True, 3]
print(mixed)

empty = []                 # an empty list (we'll fill it later)
print(empty, len(empty))
````

## 3.2 Indexing: accessing items
Every item has a **position number (index)**. **Counting starts at 0!**

```
drug_names = [ "Aspirin", "Penicillin", "Paracetamol", "Rifampicin", "Ethanol" ]
index:             0           1             2              3            4
negative:         -5          -4            -3             -2           -1
```

````python
print(drug_names[0])      # first item
print(drug_names[2])      # third item
print(drug_names[-1])     # last item
print(drug_names[-2])     # second to last
````

Going past the end gives an **IndexError**. Let's see it (and learn to read error messages!):

````python
try:
    print(drug_names[10])
except IndexError as e:
    print("⚠️ IndexError:", e)
````

## 3.3 Slicing: getting a portion
`list[start:stop]` gives items from `start` up to **but not including** `stop`. (Same rule as `range`!)

````python
print(drug_names[1:4])     # items at index 1, 2, 3
print(drug_names[:3])      # from the beginning up to index 3 (excluded)
print(drug_names[2:])      # from index 2 to the end
print(drug_names[::2])     # every 2nd item
print(drug_names[::-1])    # reversed!
````

## 3.4 Lists are mutable (changeable)

````python
numbers = [10, 20, 30]
numbers[1] = 99            # change an item
print(numbers)
````

## 3.5 Essential list methods

| Method | What it does |
|---|---|
| `lst.append(x)` | Add `x` to the **end** |
| `lst.insert(i, x)` | Insert `x` at position `i` |
| `lst.remove(x)` | Remove the **first** occurrence of value `x` |
| `lst.pop()` | Remove and **return** the last item (or `pop(i)` for index `i`) |
| `lst.sort()` | Sort the list **in place** |
| `lst.reverse()` | Reverse in place |
| `lst.count(x)` | How many times `x` appears |
| `lst.index(x)` | Position of first `x` |
| `x in lst` | `True` if `x` is in the list |

````python
fruits = ["apple", "banana"]

fruits.append("cherry")
print("After append:", fruits)

fruits.insert(1, "avocado")
print("After insert:", fruits)

fruits.remove("banana")
print("After remove:", fruits)

last = fruits.pop()
print("Popped:", last, "| Remaining:", fruits)

print("'apple' in fruits?", "apple" in fruits)
print("'mango' in fruits?", "mango" in fruits)
````

````python
values = [5, 2, 9, 1, 7]
values.sort()
print("Ascending: ", values)

values.sort(reverse=True)
print("Descending:", values)
````

### Useful built-in functions on lists

## 3.6 Lists + loops = 💪
### Three ways to loop over a list

| Style | Use when |
|---|---|
| `for item in lst` | You only need the **values** (most common) |
| `for i in range(len(lst))` | You need the **index** |
| `for i, item in enumerate(lst)` | You need **both** (cleanest) |

````python
# Style 1: values only
for name in drug_names:
    print(name)
````

````python
# Style 2: using the index
for i in range(len(drug_names)):
    print(i, drug_names[i], molecular_weights[i])
````

````python
# Style 3: enumerate gives (index, value) pairs
for i, name in enumerate(drug_names):
    print(f"{i + 1}. {name} (MW = {molecular_weights[i]})")
````

````python
# zip() walks through two lists in parallel
for name, mw in zip(drug_names, molecular_weights):
    print(f"{name:12} {mw:7.1f}")
````

## 3.7 Building lists in loops

### Pattern A: Build a list
Start with an empty list and `append` in a loop.

````python
squares = []
for n in range(1, 8):
    squares.append(n ** 2)
print(squares)
````

### Pattern B: Filter a list
Keep only the items that pass a test.

````python
# Which drugs satisfy the Lipinski MW rule (MW <= 500)?
light_drugs = []
for name, mw in zip(drug_names, molecular_weights):
    if mw <= 500:
        light_drugs.append(name)

print("MW <= 500:", light_drugs)
````

### Pattern C: Transform a list
Convert every item (e.g., Daltons → kilodaltons).

````python
mw_kda = []
for mw in molecular_weights:
    mw_kda.append(mw / 1000)
print(mw_kda)
````

### 🎁 Bonus: list comprehensions
A compact one-line version of "build/filter/transform". Don't worry if it looks strange at first; the loops above are perfectly fine!

````python
squares = [n ** 2 for n in range(1, 8)]
light_drugs = [name for name, mw in zip(drug_names, molecular_weights) if mw <= 500]
mw_kda = [mw / 1000 for mw in molecular_weights]

print(squares)
print(light_drugs)
print(mw_kda)
````

## 3.8 ⚠️ Common list pitfalls

**1. Don't modify a list while looping over it** (items get skipped!)

````python
nums = [2, 4, 6, 8, 9]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)        # ❌ dangerous
print(nums)                   # we wanted only the odd numbers: [9]. What did we get?
````

````python
# ✅ Safe way: build a NEW list
nums = [2, 4, 6, 8, 9]
odds = [n for n in nums if n % 2 != 0]
print(odds)
````

**2. Copying a list: `b = a` does NOT copy!** Both names point to the *same* list.

````python
a = [1, 2, 3]
b = a                # NOT a copy, just another name for the same list
b.append(99)
print("a =", a)       # a changed too!

c = a.copy()         # ✅ a real, independent copy
c.append(100)
print("a =", a)
print("c =", c)
````

# 📕 Session 4: Dictionaries

## 4.1 What is a dictionary?
A list finds items by **position** (0, 1, 2...). A **dictionary** finds items by a **key**, like a real dictionary (look up a *word* to get its *meaning*) or a codon table (look up a *codon* to get its *amino acid*).

```python
my_dict = {key1: value1, key2: value2, ...}
```

| List | Dictionary |
|---|---|
| `[ ]` square brackets | `{ }` curly braces |
| Access by **position**: `lst[0]` | Access by **key**: `d["Aspirin"]` |
| Ordered sequence of values | Collection of **key → value** pairs |

````python
molecular_weight = {
    "Aspirin": 180.2,
    "Penicillin": 342.3,
    "Paracetamol": 151.2
}

print(molecular_weight)
print("Aspirin MW:", molecular_weight["Aspirin"])
print("Number of entries:", len(molecular_weight))
````

## 4.2 Working with dictionaries

| Operation | Syntax |
|---|---|
| Get a value | `d[key]` or `d.get(key)` |
| Add / update | `d[key] = value` |
| Delete | `del d[key]` or `d.pop(key)` |
| Check key exists | `key in d` |
| All keys | `d.keys()` |
| All values | `d.values()` |
| All pairs | `d.items()` |

🔑 **Keys must be unique** (and unchangeable, e.g. strings, numbers). **Values can be anything.**

````python
mw = {"Aspirin": 180.2, "Penicillin": 342.3}

mw["Ethanol"] = 46.1          # add a new entry
mw["Aspirin"] = 180.16        # update an existing entry
print(mw)

del mw["Penicillin"]          # delete
print(mw)

print("Is Aspirin in mw?", "Aspirin" in mw)
print("Is Caffeine in mw?", "Caffeine" in mw)
````

### `KeyError` and the safe alternative `.get()`

````python
try:
    print(mw["Caffeine"])
except KeyError as e:
    print("⚠️ KeyError: key not found →", e)

# .get() returns None (or a default you choose) instead of crashing
print(mw.get("Caffeine"))
print(mw.get("Caffeine", "not in database"))
````

## 4.3 Looping over a dictionary

````python
mw = {"Aspirin": 180.2, "Penicillin": 342.3, "Paracetamol": 151.2, "Rifampicin": 780.9}

# Loop over keys (the default)
for drug in mw:
    print(drug)
````

````python
# Loop over values
for value in mw.values():
    print(value)
````

````python
# Loop over both: the most useful form!
for drug, weight in mw.items():
    print(f"{drug:12} → {weight} Da")
````

## 4.4 The counting pattern 📊
**Counting how often things occur** is perhaps the most useful thing dictionaries do. The idea: the *key* is the thing you're counting, the *value* is the count.

````python
dna = "ATGCGATACGCTTAGGA"

counts = {}                       # start with an empty dictionary
for base in dna:
    if base in counts:
        counts[base] += 1         # seen before → add 1
    else:
        counts[base] = 1          # first time → start at 1

print(counts)
````

A neater version using `.get()` with a default of 0:

````python
counts = {}
for base in dna:
    counts[base] = counts.get(base, 0) + 1
print(counts)
````

# 🏆 Capstone Mini-Project: Lipinski's Rule of Five

Time to combine **loops, lists, dictionaries, conditions and counters**.

**Lipinski's Rule of Five** predicts whether a molecule is likely to be orally bioavailable. A compound is **compliant** if it satisfies **all** of:

| Property | Criterion |
|---|---|
| Molecular weight (MW) | ≤ 500 Da |
| XlogP (lipophilicity) | ≤ 5 |
| H-bond donors (HBD) | ≤ 5 |
| H-bond acceptors (HBA) | ≤ 10 |

We'll use a small dataset of approximate values for well-known drugs. (Values are for **teaching purposes** and approximate; real analyses use databases like e-Drug3D with tools like DataWarrior.)

````python
library = [
    {"name": "Aspirin",        "MW": 180.2,  "XlogP": 1.2, "HBD": 1, "HBA": 4},
    {"name": "Paracetamol",    "MW": 151.2,  "XlogP": 0.5, "HBD": 2, "HBA": 3},
    {"name": "Ibuprofen",      "MW": 206.3,  "XlogP": 3.5, "HBD": 1, "HBA": 2},
    {"name": "Caffeine",       "MW": 194.2,  "XlogP": -0.1,"HBD": 0, "HBA": 6},
    {"name": "Metformin",      "MW": 129.2,  "XlogP": -1.4,"HBD": 3, "HBA": 3},
    {"name": "Atenolol",       "MW": 266.3,  "XlogP": 0.2, "HBD": 3, "HBA": 4},
    {"name": "Diazepam",       "MW": 284.7,  "XlogP": 2.8, "HBD": 0, "HBA": 3},
    {"name": "Rifampicin",     "MW": 822.9,  "XlogP": 3.7, "HBD": 6, "HBA": 16},
    {"name": "Cyclosporine",   "MW": 1202.6, "XlogP": 3.0, "HBD": 5, "HBA": 12},
    {"name": "Amiodarone",     "MW": 645.3,  "XlogP": 7.6, "HBD": 0, "HBA": 5},
    {"name": "Vancomycin",     "MW": 1449.3, "XlogP": -3.1,"HBD": 18,"HBA": 24},
    {"name": "Atorvastatin",   "MW": 558.6,  "XlogP": 5.7, "HBD": 4, "HBA": 7},
]
print(f"Library contains {len(library)} compounds.")
````

## Step 1: Write the rule check
Loop through every compound, check each criterion, and record **which rules it breaks**.

````python
compliant = []        # list of compliant drug dictionaries
violations = {}       # dictionary: drug name → list of broken rules

for drug in library:
    broken = []
    if drug["MW"] > 500:
        broken.append("MW > 500")
    if drug["XlogP"] > 5:
        broken.append("XlogP > 5")
    if drug["HBD"] > 5:
        broken.append("HBD > 5")
    if drug["HBA"] > 10:
        broken.append("HBA > 10")

    if len(broken) == 0:
        compliant.append(drug)
    else:
        violations[drug["name"]] = broken

print("✅ Compliant   :", [d["name"] for d in compliant])
print("❌ Violators   :")
for name, rules in violations.items():
    print(f"   {name:14} breaks → {', '.join(rules)}")
````

## Step 2: How many compounds comply?

````python
total = len(library)
n_ok = len(compliant)
percent = n_ok / total * 100

print(f"{n_ok} out of {total} compounds satisfy all four criteria ({percent:.1f}%).")
````

## Step 3: Which rule is broken most often?
This is the **counting pattern** from Session 4.

````python
rule_counts = {"MW > 500": 0, "XlogP > 5": 0, "HBD > 5": 0, "HBA > 10": 0}

for rules in violations.values():
    for rule in rules:
        rule_counts[rule] += 1

for rule, n in rule_counts.items():
    print(f"{rule:10} {'█' * n} {n}")
````

## Step 4: Summary statistics for the compliant set

````python
mw_values = [d["MW"] for d in compliant]
print(f"Compliant MW range : {min(mw_values)} – {max(mw_values)} Da")
print(f"Compliant mean MW  : {sum(mw_values) / len(mw_values):.1f} Da")
````

## Step 5 (optional): Make a plot 📈
Colab includes `matplotlib` for plotting. We'll plot MW against XlogP and colour compliant vs non-compliant drugs.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))

for drug in library:
    ok = drug["name"] not in violations
    ax.scatter(drug["MW"], drug["XlogP"],
               color="green" if ok else "red", s=80)
    ax.annotate(drug["name"], (drug["MW"], drug["XlogP"]),
                textcoords="offset points", xytext=(5, 5), fontsize=8)

ax.axvline(500, linestyle="--", color="grey")   # MW limit
ax.axhline(5, linestyle="--", color="grey")     # XlogP limit
ax.set_xlabel("Molecular weight (Da)")
ax.set_ylabel("XlogP")
ax.set_title("Lipinski space (green = compliant, red = violator)")
plt.show()
````

# 📝 Summary Cheat-Sheet

### Loops
```python
while condition:          # repeat while True; don't forget to update!
    ...

for item in sequence:     # repeat for each item
    ...

for i in range(start, stop, step):   # stop is EXCLUDED
    ...
```
`break` → exit loop  |  `continue` → skip to next pass

### Lists `[ ]`
```python
lst = [10, 20, 30]
lst[0]            # indexing (starts at 0)
lst[-1]           # last item
lst[1:3]          # slicing (stop excluded)
lst.append(40)    # add to end
lst.remove(20)    # remove a value
lst.sort()        # sort in place
len(lst), sum(lst), min(lst), max(lst)
for i, x in enumerate(lst): ...
```

### Dictionaries `{ }`
```python
d = {"a": 1, "b": 2}
d["c"] = 3                    # add / update
d["a"]                        # lookup (KeyError if missing)
d.get("z", 0)                 # safe lookup with default
"a" in d                      # key exists?
for k, v in d.items(): ...    # loop over pairs
```

### Patterns to remember
| Pattern | Idea |
|---|---|
| **Accumulator** | `total = 0` … `total += x` |
| **Counter** | `count += 1` inside an `if` |
| **Filter** | build a new list with only items passing a test |
| **Counting with dict** | `d[x] = d.get(x, 0) + 1` |
| **List of dicts** | a small table: each dict is a row |

### Common errors & what they mean
| Error | Usually means |
|---|---|
| `IndentationError` | Wrong or missing indentation |
| `IndexError` | List index is out of range |
| `KeyError` | Dictionary key doesn't exist |
| `TypeError` | Mixing incompatible types (e.g., `"5" + 3`) |
| `NameError` | Using a variable that hasn't been defined |
| *(cell runs forever)* | Infinite loop; press ⏹️ stop |
