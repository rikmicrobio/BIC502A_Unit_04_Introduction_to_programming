## 1. What is a conditional statement?

A conditional statement allows a program to make a decision. A condition evaluates to `True` or `False`.

**Condition → Decision → Action**

````python
print(True)
print(False)

x = 10
print(x > 5)
print(x == 10)
````

## 2. The `if` statement

### Syntax
```python
if condition:
    statement
```
The colon starts the block and indentation identifies the statements belonging to the `if`.

````python
temperature = 37

if temperature == 37:
    print("Temperature is 37°C")
````

### Biological example

````python
pH = 8.2

if pH > 7:
    print("The solution is basic.")
````

### Try it
Change `pH` to `7`, `7.5`, and `6.5`. Observe when the message appears.

## 3. The `if–else` statement

Use `if–else` when there are two alternatives.

```python
if condition:
    ...
else:
    ...
```

````python
temperature = 39

if temperature <= 37:
    print("Normal temperature")
else:
    print("High temperature")
````

### Example: DNA sequence length

````python
sequence_length = 85

if sequence_length >= 100:
    print("Long sequence")
else:
    print("Short sequence")
````

## 4. The `if–elif–else` statement

Use `elif` when there are more than two possibilities. Python checks conditions from top to bottom and executes the first true branch.

````python
score = 82

if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Grade D")
````

### Biological examples

````python
pH = 7.4

if pH < 7:
    print("Acidic")
elif pH == 7:
    print("Neutral")
else:
    print("Basic")
````

````python
temperature = 37

if temperature < 35:
    print("Low")
elif temperature <= 37:
    print("Normal")
else:
    print("High")
````

## 5. Comparison operators in conditions

`==` equal, `!=` not equal, `>` greater, `<` less, `>=` greater/equal, `<=` less/equal.

````python
x = 25

print(x >= 20)
print(x == 25)
print(x != 25)
print(x < 10)
````

## 6. Logical operators

Logical operators combine Boolean conditions.

| Operator | Meaning |
|---|---|
| `and` | both conditions must be True |
| `or` | at least one condition must be True |
| `not` | reverses True/False |

## 7. `and`

`and` is True only when **both** conditions are True.

````python
age = 20
has_id = True

if age >= 18 and has_id:
    print("Allowed")
````

### Biological example

````python
pH = 7.2
temperature = 37

if pH >= 7.0 and pH <= 7.5 and temperature == 37:
    print("Sample conditions are suitable")
````

## 8. `or`

`or` is True when **at least one** condition is True.

````python
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("Weekend")
````

### Biological example: purines

````python
base = "G"

if base == "A" or base == "G":
    print("Purine")
else:
    print("Pyrimidine")
````

## 9. `not`

`not` reverses a Boolean value: `not True` → `False`; `not False` → `True`.

````python
contaminated = False

if not contaminated:
    print("Sample can proceed")
````

## 10. Combining logical operators

````python
sample_available = True
contaminated = False
concentration = 15

if sample_available and not contaminated and concentration >= 10:
    print("Sample accepted")
else:
    print("Sample rejected")
````

## 11. Nested conditions vs logical operators

The same simple decision can sometimes be written with nesting or with `and`.

````python
sample_available = True
concentration = 15

# Nested version
if sample_available:
    if concentration >= 10:
        print("Sample accepted")

# Using and
if sample_available and concentration >= 10:
    print("Sample accepted")
````

## 12. Complete biological example

A sample is accepted only if it is available, not contaminated, pH is 7.0–7.5, temperature is 37°C, and concentration is at least 10.

````python
sample_available = True
contaminated = False
pH = 7.2
temperature = 37
concentration = 15

if sample_available and not contaminated:
    if pH >= 7.0 and pH <= 7.5 and temperature == 37:
        if concentration >= 10:
            print("Sample accepted")
        else:
            print("Concentration too low")
    else:
        print("Experimental conditions are unsuitable")
else:
    print("Sample unavailable or contaminated")
````

## 13. Common beginner mistakes

1. `=` is assignment; `==` is comparison.
2. Remember the colon `:` after `if`, `elif`, and `else`.
3. Indent statements inside a conditional block.
4. Use `and` when **all** conditions are required.
5. Use `or` when **at least one** condition is enough.
6. Use `not` to reverse a Boolean condition.

# Final takeaway

**Conditional statements control program flow.**

- `if` → one condition
- `if–else` → two alternatives
- `if–elif–else` → multiple alternatives
- `and` → all conditions must be true
- `or` → at least one condition must be true
- `not` → reverses a Boolean condition

> **Condition → Decision → Action**
